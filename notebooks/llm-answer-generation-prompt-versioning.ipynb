```python
import instructor
from pydantic import BaseModel, Field
from typing import Optional
from typing import List
from instructor import Mode
```

```python
client = instructor.from_provider(
    "ollama/llama3",
    base_url="http://localhost:11434/v1",
    mode=Mode.JSON   
    )
```

```python
class RAGRetrievedContext(BaseModel):
    id: str = Field(description="id of the game chosen")
    description: str = Field(description = "short description of the game used to answer the question")
    name: str= Field(description="Name of the game chosen")
```

```python
def generate_answer(prompt):
    response = client.create(
    model="llama3",
    messages=[
        {
            "role": "user",
            "content": prompt,
        }
    ],
    response_model=RAGRetrievedContext,
    mode=Mode.JSON
)

    return response.model_dump_json()
    
```

```python
def process_context(context):
    formatted = ""

    for i in range(len(context["retrieved_context_ids"])):
        formatted += f"""
ID: {context["retrieved_context_ids"][i]}
Name: {context["retrieved_names"][i]}
Genre: {context["retrieved_genres"][i]}
Score: {context["similarity_scores"][i]}
Description: {context["retrieved_contexts"][i][:300]}
---
"""
    return formatted
```
