##  Basic analysis and filtering data

### Importing Dependencies

```python
import pandas as pd
```

### Analyse CSV Files

```python
df_steam = pd.read_csv("../raw data/steam.csv")
```

```python
df_steam.head()
```

```python
is_duplicated = not df_steam["appid"].is_unique
print(is_duplicated)
```

```python
df_steam_description = pd.read_csv("../raw data/steam_description_data.csv")
df_steam_media = pd.read_csv("../raw data/steam_media_data.csv")
df_steam_requirements = pd.read_csv("../raw data/steam_requirements_data.csv")
df_steam_support_info = pd.read_csv("../raw data/steam_support_info.csv")
df_steam_spytag = pd.read_csv("../raw data/steamspy_tag_data.csv")
```

```python
df_steam_requirements.head()
print(df_steam_requirements["pc_requirements"][0])
```

```python
df_steam_master = pd.merge(df_steam,df_steam_requirements, left_on="appid", right_on="steam_appid", how="inner")
```

```python
df_steam_master.head()
```

### Check if all the remaining lists have steam_appid or appid

```python
print("steam_appid" in df_steam_description )
print("steam_appid" in df_steam_media )
print("steam_appid" in df_steam_support_info )
print("steam_appid" in df_steam_spytag )
print("appid" in df_steam_spytag )

```

### Join all DFs to master DF

```python
df_steam_master = df_steam_master.merge(df_steam_description, on="steam_appid").merge(df_steam_media, on="steam_appid").merge(df_steam_support_info, on="steam_appid").merge(df_steam_spytag, on="appid")
```
